En este ejercicio se generará un conjunto de transacciones simuladas para identificar posibles casos de fraude según la hora del día y el monto de la operación.
</p>

</div>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

<div style="background:#fff7ed; border-left:5px solid #fdba74; padding:14px; border-radius:10px;">

<h3 style="color:#9a6b4f;"> Importación de librerías</h3>

<p style="color:#626b78;">
Se importan las librerías necesarias para trabajar con datos numéricos, estructuras tabulares y gráficos.
</p>

<ul style="color:#626b78;">
<li><b>NumPy:</b> permite generar números aleatorios y realizar operaciones matemáticas.</li>
<li><b>Pandas:</b> permite organizar las transacciones en un DataFrame.</li>
<li><b>Matplotlib:</b> permite representar gráficamente los datos.</li>
</ul>

</div>

```python
np.random.seed(42)

n = 300
```

<div style="background:#eff6ff; border-left:5px solid #93c5fd; padding:14px; border-radius:10px;">

<h3 style="color:#5b7fa3;">Configuración inicial</h3>

<p style="color:#5f6b7a;">
Se utiliza una semilla aleatoria con valor <b>42</b> para garantizar que los resultados generados sean reproducibles.
</p>

<p style="color:#5f6b7a;">
La variable <b>n</b> indica que se crearán <b>300 transacciones simuladas</b>.
</p>

</div>

```python
hora = np.random.randint(0, 24, n)

monto = np.random.exponential(scale=150, size=n) + 10
```

<div style="background:#ecfdf5; border-left:5px solid #86efac; padding:14px; border-radius:10px;">

<h3 style="color:#4f7c68;">Generación de transacciones</h3>

<p style="color:#5f6b7a;">
Se generan dos características para cada transacción:
</p>

<ul style="color:#5f6b7a;">
<li><b>hora:</b> número entero entre 0 y 23 que representa la hora del día.</li>
<li><b>monto:</b> cantidad de dinero de la transacción.</li>
</ul>

<p style="color:#5f6b7a;">
Los montos se generan utilizando una distribución exponencial, lo que permite obtener muchas operaciones de montos pequeños y algunas operaciones de montos elevados.
</p>

</div>

```python
prob_fraude = (
    0.03
    + 0.35 * ((hora <= 5) | (hora >= 23))
    + 0.25 * (monto > 400)
)

prob_fraude = np.clip(prob_fraude, 0, 0.9)
```

<div style="background:#fff1f2; border-left:5px solid #fda4af; padding:14px; border-radius:10px;">

<h3 style="color:#a65f69;">Probabilidad de fraude</h3>

<p style="color:#5f6b7a;">
Se define una probabilidad base de fraude de <b>3%</b>.
</p>

<p style="color:#5f6b7a;">
La probabilidad aumenta cuando se presentan ciertas condiciones:
</p>

<ul style="color:#5f6b7a;">
<li>Se incrementa en <b>35%</b> si la operación ocurre entre las 00:00 y 05:00 horas o después de las 23:00.</li>
<li>Se incrementa en <b>25%</b> si el monto supera los <b>S/ 400</b>.</li>
</ul>

<p style="color:#5f6b7a;">
Finalmente, <b>np.clip()</b> limita la probabilidad máxima a 0.9, es decir, 90%.
</p>

</div>

```python
es_fraude = (np.random.rand(n) < prob_fraude).astype(int)
```

<div style="background:#f5f3ff; border-left:5px solid #c4b5fd; padding:14px; border-radius:10px;">

<h3 style="color:#7665a8;">Clasificación de las transacciones</h3>

<p style="color:#5f6b7a;">
Se genera un número aleatorio entre 0 y 1 para cada transacción.
</p>

<p style="color:#5f6b7a;">
Si dicho valor es menor que la probabilidad de fraude calculada, la transacción se clasifica como fraude.
</p>

<ul style="color:#5f6b7a;">
<li><b>0:</b> transacción legítima.</li>
<li><b>1:</b> transacción fraudulenta.</li>
</ul>

</div>

```python
transacciones = pd.DataFrame({
    "hora": hora,
    "monto": monto,
    "es_fraude": es_fraude
})

transacciones.head()
```
